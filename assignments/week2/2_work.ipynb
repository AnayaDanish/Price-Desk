```python
clean = raw.copy()
```

```python
print("original shape:", raw.shape)
print("clean shape:", clean.shape)
```

```python
print("missing before", clean["price"].isna().sum())
```

```python
clean["price"] = clean["price"].str.replace(",", "", regex=False)
```

```python
clean["price"] = pd.to_numeric(clean["price"], errors="coerce")
```

```python
print("price type:", clean["price"].dtype)
print("missing after:", clean["price"].isna().sum())
```

```python
clean["date"] = pd.to_datetime(clean["date"], errors="coerce")
```

```python
print("date type:", clean["date"].dtype)
print("missing dates:", clean["date"].isna().sum())
```

```python
print("duplicates before:", clean.duplicated().sum())
```

```python
clean[clean.duplicated(keep=False)]
```

```python
clean = clean.drop_duplicates()
```

```python
print("duplicates after:", clean.duplicated().sum())
print("new shape:", clean.shape)
```

```python
clean[clean["price"] == 9999]
```

```python
clean["price_flag"] = clean["price"] == 9999

print(clean[clean["price_flag"]])
```

```python
clean = clean.sort_values("date")

print(clean[["date", "item", "city", "price"]].head())
```

```python
print("final shape:", clean.shape)
print("missing prices:", clean["price"].isna().sum())
print("duplicates:", clean.duplicated().sum())
print("date type:", clean["date"].dtype)
print("price type:", clean["price"].dtype)
```

```python
clean.to_csv("teaching_prices_cleaned.csv", index=False)

print("cleaned dataset saved")
```
